```python
import pandas as pd
import numpy as np

# ============================================
# 0. 파일 불러오기
# ============================================
path = r"C:\Users\JEON YUNA\Downloads\archive (1)\amazon.csv"
df = pd.read_csv(path)

print("원본 shape:", df.shape)

# ============================================
# 1. discounted_price, actual_price
#    ₹ 기호, 콤마(,) 제거 후 float으로 변환
# ============================================
for col in ['discounted_price', 'actual_price']:
    df[col] = (
        df[col]
        .str.replace('₹', '', regex=False)
        .str.replace(',', '', regex=False)
        .astype(float)
    )

# ============================================
# 2. discount_percentage
#    % 기호 제거 후 float으로 변환
# ============================================
df['discount_percentage'] = (
    df['discount_percentage']
    .str.replace('%', '', regex=False)
    .astype(float)
)

# ============================================
# 3. rating_count
#    콤마(,) 제거 후 float으로 변환 (결측치 2개는 NaN 유지)
# ============================================
df['rating_count'] = (
    df['rating_count']
    .str.replace(',', '', regex=False)
    .astype(float)
)

# ============================================
# 4. rating
#    '|' 같은 숫자가 아닌 값은 NaN 처리 후 float으로 변환
# ============================================
df['rating'] = pd.to_numeric(df['rating'], errors='coerce')

# ============================================
# (5번 완전/부분 중복 행 제거는 이번 단계에서 보류 - 나중에 별도 결정)
# ============================================

# ============================================
# 결과 확인
# ============================================
print("\n전처리 후 데이터 타입:")
print(df[['discounted_price', 'actual_price', 'discount_percentage',
          'rating', 'rating_count']].dtypes)

print("\n전처리 후 결측치:")
print(df[['discounted_price', 'actual_price', 'discount_percentage',
          'rating', 'rating_count']].isnull().sum())

print("\n샘플 확인 (상위 5행):")
print(df[['product_id', 'discounted_price', 'actual_price',
          'discount_percentage', 'rating', 'rating_count']].head())

# rating이 NaN으로 바뀐 행 확인 (원래 '|' 였던 행)
print("\nrating이 NaN인 행:")
print(df[df['rating'].isnull()][['product_id', 'product_name']])

# ============================================
# CSV로 저장
# ============================================
out_path = r"C:\Users\JEON YUNA\Downloads\archive (1)\amazon_cleaned.csv"
df.to_csv(out_path, index=False)
print(f"\n저장 완료: {out_path}")
```
